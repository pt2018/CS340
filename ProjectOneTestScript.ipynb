```python
from pprint import pprint
from CRUD_Python_Module import AnimalShelter
```

```python
# Supply the credentials required by the uploaded CRUD constructor.
shelter = AnimalShelter("aacuser", "SNHU1234")
print("AnimalShelter object created.")
```

```python
animal_id = shelter.getNextAnimalID()

test_animal = {
    "animal_id": animal_id,
    "name": "Buddy",
    "animal_type": "Dog",
    "breed": "Labrador Retriever Mix",
    "age_upon_outcome": "2 years",
    "sex_upon_outcome": "Neutered Male",
    "outcome_type": "Adoption"
}

create_result = shelter.create(test_animal)

print("Create result:", create_result)
print("Animal ID:", animal_id)

assert create_result is True, "Create test failed."
print("Create test passed.")
```

```python
read_result = shelter.read({"animal_id": animal_id})

pprint(read_result)

assert isinstance(read_result, list), "Read must return a list."

# Identify this inserted document by MongoDB's unique _id.
matches = [
    animal for animal in read_result
    if animal.get("name") == "Buddy"
]
assert len(matches) == 1, "Expected one matching test document."

test_query = {"_id": matches[0]["_id"]}

print("Read test passed.")
```

```python
update_result = shelter.update(
    test_query,
    {"$set": {"name": "Buddy Updated"}}
)

print("Documents modified:", update_result)
assert update_result == 1, "Update count was incorrect."

# Read the document again to verify the stored change.
updated_animal = shelter.read(test_query)
pprint(updated_animal)

assert len(updated_animal) == 1, "Updated document was not found."
assert updated_animal[0]["name"] == "Buddy Updated", "Name was not updated."

print("Update test passed.")
```

```python
delete_result = shelter.delete(test_query)

print("Documents deleted:", delete_result)
assert delete_result == 1, "Delete count was incorrect."

# Verify that the deleted document is no longer returned.
remaining_animals = shelter.read(test_query)

print("Read after delete:", remaining_animals)
assert remaining_animals == [], "Document still exists."

print("Delete test passed.")
print("All CRUD tests passed.")
```
